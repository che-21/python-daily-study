```python
# 0. 절댓값 구하기 알고리즘
import math

# 1. 부호 판단
def abs_sign(a):
    if a >= 0:
        return a
    else:
        return -a

# 2. 제곱-제곱근
def abs_squares(a):
    b = a * a
    return math.sqrt(b) #수학 모듈의 제곱근 함수

print(abs_sign(5))
print(abs_sign(-3))
print()
print(abs_squares(5))
print(abs_squares(-3))
```

```python
# 1. 1부터 n까지의 합 구하기
# 1-1. 1부터 n까지 연속한 숫자의 합을 구하는 알고리즘1
def sum_n(n):
    s = 0
    for i in range(1, n + 1):
        s = s + i
    return s

print(sum_n(10))
print(sum_n(100))

# 1-2. 1부터 n까지 연속한 숫자의 합을 구하는 알고리즘2
def sum_n(n):
    return n * (n + 1) // 2

print(sum_n(10))
print(sum_n(100))
```

```python
# 2. 최댓값  찾기
# 2-1. 최댓값 찾기1
def find_max(a):
    n = len(a)
    max_v = a[0]
    for i in range(1, n):
        if a[i] > max_v:
            max_v = a[i]
    return max_v

v = [17, 92, 18, 33, 58, 7, 33, 42]
print(find_max(v))

# 2-2. 최댓값 찾기2 - 최댓값의 "위치"를 구하는 알고리즘
def find_max_idx(a):
    n = len(a)
    max_idx = 0
    for i in range(1, n):
        if a[i] > a[max_idx]:
            max_idx = i
    return max_idx

v = [17, 92, 18, 33, 58, 7, 33, 42]
print(find_max_idx(v))
```

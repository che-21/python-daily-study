```python
# 06-2 3과 5의 배수를 모두 더하기
"""
10 미만의 자연수에서 3과 5의 배수를 구하면 3, 5, 6, 9이다. 이들의 총합은 23이다.
1,000 미만의 자연수에서 3의 배수와 5의 배수의 총합을 구하라.
"""
result = 0
for n in range(1, 1000):
    # 3과 5의 공배수 15를 걸러내기 위해 or 연산자 사용
    if n % 3 == 0 or n % 5 == 0:
        result += n
print(result)
```

```python
# 06-3 게시판 페이징 하기
"""
게시물의 총 개수가 5이고 한 페이지에 보여 줄 게시물 수가 10이면 총 페이지 수는 당연히 1이 된다.
만약 게시물의 총 개수가 15이고 한 페이지에 보여 줄 게시물 수가 10이라면 총 페이지 수는 2가 될 것이다.
"""
# m: 게시물 수
# n: 한 페이지에 보여줄 게시물 수
def get_total_page(m,n):
    if m%n == 0:
        return m//n
    else:
        return m//n+1

print(get_total_page(5, 10))    # 1 출력
print(get_total_page(15, 10))   # 2 출력
print(get_total_page(25, 10))   # 3 출력
print(get_total_page(30, 10))   # 3 출력
```

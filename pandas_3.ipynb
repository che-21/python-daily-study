```python
# 인덱스 활용
## <예제 1-16> 특정 열을 행 인덱스로 설정
import pandas as pd

# DataFrame() 함수로 데이터프레임 변환, 변수 df에 저장
exam_data = {'이름': ['서준', '우현', '인아'],
             '수학': [90, 80, 70],
             '영어': [98, 89, 95],
             '음악': [85, 95, 100],
             '체육': [100, 90, 90]}
df = pd.DataFrame(exam_data)

# 특정 열(column)을 데이터프레임의 행 인덱스(index)로 설정
ndf = df.set_index(['이름'])
print(ndf)
print('\n')

ndf2 = ndf.set_index('음악')
print(ndf2)
print('\n')

ndf3 = ndf.set_index(['수학', '음악'])
print(ndf3)
```

```python
# 행 인덱스 재배열
## <예제 1-17> 새로운 배열로 행 인덱스를 재지정
import pandas as pd

# 딕셔너리를 데이터프레임으로 변환
dict_data = {'c0':[1,2,3], 'c1':[4,5,6], 'c2':[7,8,9], 'c3':[10,11,12], 'c4':[13,14,15]}

# 인덱스를 [r0, r1, r2]로 지정
df = pd.DataFrame(dict_data, index=['r0', 'r1', 'r2'])
print(df)
print('\n')

# 인덱스를 [r0, r1, r2, r3, r4]로 재지정
"""기존 데이터프레임 df에는 r3, 34가 없었기 때문에 값들이 결측치(NaN)로 채워지는데,
NumPy와 판다스의 기본 정수형 타입은 NaN을 표현할 수 없어
정수형 전체를 실수형으로 자동변환해 NaN을 수용할 수 있도록 함"""

new_index = ['r0', 'r1', 'r2', 'r3', 'r4']
ndf = df.reindex(new_index)
print(ndf)
print('\n')

# reindex로 발생한 NaN 값을 숫자 0으로 채우기
new_index = ['r0', 'r1', 'r2', 'r3', 'r4']
ndf2 = df.reindex(new_index, fill_value=0)
print(ndf2)
```

```python
# 행 인덱스 초기화
## <예제 1-18> 정수형 위치 인덱스로 초기화
import pandas as pd

# 딕셔너리를 데이터프레임으로 변환
dict_data = {'c0':[1,2,3], 'c1':[4,5,6], 'c2':[7,8,9], 'c3':[10,11,12], 'c4':[13,14,15]}

# 인덱스를 [r0, r1, r2]로 지정
df = pd.DataFrame(dict_data, index=['r0', 'r1', 'r2'])
print(df)
print('\n')

# 행 인덱스를 정수형으로 초기화
ndf = df.reset_index()
print(ndf)
print('\n')

# 행 인덱스를 정수형으로 초기화하고, 기존 인덱스의 열 이름을 지정
ndf2 = df.reset_index(names=['C00'])
print(ndf2)
print('\n')

# 행 인덱스를 초기화하고, 기존인덱스를 삭제
ndf3 = df.reset_index(drop=True)
print(ndf3)
```

```python
# 행 인덱스를 기준으로 데이터프레임 정렬
## <예제 1-19> 데이터프레임 정렬
import pandas as pd

# 딕셔너리를 데이터 프레임으로 변환
dict_data = {'c0':[1,2,3], 'c1':[4,5,6], 'c2':[7,8,9], 'c3':[10,11,12], 'c4':[13,14,15]}

# 인덱스를 [r0, r1, r2]로 지정
df = pd.DataFrame(dict_data, index=['r0', 'r1', 'r2'])
print(df)
print('\n')

# 내림차순으로 행 인덱스 정렬
ndf = df.sort_index(ascending=False)
print(ndf)
print('\n')

# 오름차순으로 행 인덱스 정렬
ndf2 = df.sort_index(ascending=True)
print(ndf2)
```

```python
# 특정 열의 데이터 값을 기준으로 데이터 프레임 정렬
## <예제 1-20> 열 기준 정렬
import pandas as pd

# 딕셔너리를 데이터 프레임으로 변환
dict_data = {'c0':[1,2,3], 'c1':[4,5,6], 'c2':[7,8,9], 'c3':[10,11,12], 'c4':[13,14,15]}

# 인덱스를 [r0, r1, r2]로 지정
df = pd.DataFrame(dict_data, index=['r0', 'r1', 'r2'])
print(df)
print('\n')

# c1 열을 기준으로 내림차순 정렬
ndf = df.sort_values(by='c1', ascending=False)
print(ndf)
print('\n')

# c2 열을 기준으로 오름차순 정렬
ndf2 = df.sort_values(by='c2', ascending=True)
print(ndf2)
print('\n')

# c3 열을 기준으로 내림차순 정렬, c4 열을 기준으로 오름차순 정렬
ndf3 = df.sort_values(by=['c3', 'c4'], ascending=[False, True])
print(ndf3)
```

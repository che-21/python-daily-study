```python
"""
1. 평균 점수 구하기
홍길동 씨의 과목별 점수는 다음과 같다.
홍길동 씨의 평균 점수를 구해보자.

과목 점수
국어 80
영어 75
수학 55
"""

kor = 80
eng = 75
mat = 55
avg = (kor+eng+mat)/3
print("평균 점수:", avg)
```

```python
"""
2. 홀수, 짝수 판별하기
자연수 13이 홀수인지, 짝수인지 판별할 수 있는 방법에 대해 말해보자.
"""
def oddeven(n):
    if n % 2 == 0:
        print("짝수입니다.")
    else:
        print("홀수입니다.")

oddeven(13)
```

```python
"""
3. 홍길동 씨의 주민등록번호는 881120-1068234이다.
홍길동 씨의 주민등록번호를 연월일 부분과 그 뒤의 숫자 부분으로 나누어 출력해보자
"""
pin = "881120-1068234"
yyyymmdd = pin[:6]
num = pin[7:]
print(yyyymmdd)
print(num)
```

```python
"""
4. 주민등록번호 인덱싱
주민등록번호 맨 첫 번째 숫자는 성별을 나타낸다.
주민등록번호에서 성별을 나타내는 숫자를 출력해보자.
"""
pin = "881120-1068234"
print(pin[7])
```

```python
"""
5. 문자열 바꾸기
다음과 같은 문자열 a:b:c:d가 있다.
문자열의 replace 함수를 사용하여 a#b#c#d로 바꿔 출력해보자.
"""
a = "a:b:c:d"
b = a.replace(":","#")
print(b)
```

```python
"""
6. 리스트 역순 정렬하기
[1,3,5,4,2] 리스트를 [5,4,3,2,1]로 만들어보자.
"""
a = [1,3,5,4,2]

# 아래처럼 작성 시 a.sort()는 None을 반환하기 때문에 에러 발생! 재할당이 아닌 원본을 직접 바꿀 것
#a = a.sort()
#a = a.sort(reverse=True)

a.sort()
a.sort(reverse=True)
print(a)
```

```python
"""
7. 리스트를 문자열로 만들기
['Life', 'is', 'too', 'short'] 리스트를
Life is too short 문자열로 만들어 출력해보자.
"""
a = ['Life', 'is', 'too', 'short']
result = " ".join(a)
print(result)
```

```python
"""
8. 튜플 더하기
(1,2,3) 튜플에 4를 추가하여 (1,2,3,4)를 만든 후 출력해보자.
"""
a = (1,2,3)
# .add()는 집합(Set) 자료형에 요소를 추가할 때 쓰는 메서드
# 튜플은 한 번 만들면 요소를 추가, 수정, 삭제할 수 없지만 새로운 1개짜리 튜플 (4,)를 더해서 새로운 튜플을 만들어 재할당할 수 있음
#a.add('4')
a = a + (4,)
print(a)
```

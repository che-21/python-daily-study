```python
"""
1. 평균 점수 구하기
홍길동 씨의 과목별 점수는 다음과 같다.
홍길동 씨의 평균 점수를 구해보자.

과목 점수
국어 80
영어 75
수학 55
"""

kor = 80
eng = 75
mat = 55
avg = (kor+eng+mat)/3
print("평균 점수:", avg)
```

```python
"""
2. 홀수, 짝수 판별하기
자연수 13이 홀수인지, 짝수인지 판별할 수 있는 방법에 대해 말해보자.
"""
def oddeven(n):
    if n % 2 == 0:
        print("짝수입니다.")
    else:
        print("홀수입니다.")

oddeven(13)
```

```python
"""
3. 홍길동 씨의 주민등록번호는 881120-1068234이다.
홍길동 씨의 주민등록번호를 연월일 부분과 그 뒤의 숫자 부분으로 나누어 출력해보자
"""
pin = "881120-1068234"
yyyymmdd = pin[:6]
num = pin[7:]
print(yyyymmdd)
print(num)
```

```python
"""
8. 튜플 더하기
(1,2,3) 튜플에 4를 추가하여 (1,2,3,4)를 만든 후 출력해보자.
"""
a = (1,2,3)
# .add()는 집합(Set) 자료형에 요소를 추가할 때 쓰는 메서드
# 튜플은 한 번 만들면 요소를 추가, 수정, 삭제할 수 없지만 새로운 1개짜리 튜플 (4,)를 더해서 새로운 튜플을 만들어 재할당할 수 있음
#a.add('4')
a = a + (4,)
print(a)
```

```python
"""
9. 딕셔너리의 키
다음과 같은 딕셔너리 a가 있다.
>>> a = dict()
>>> a
{}

다음 중 오류가 발생하는 경우를 고르고, 그 이유를 설명해보자.
"""

a = dict()
a['name'] = 'python'
a[('a',)] = 'python'

# 리스트([1])를 딕셔너리의 키로 사용함
# 파이썬 딕셔너리의 키는 값이 변경되지 않는 해시 가능(Hashable)하고 불변(Immutable)인 객체만 사용할 수 있음
#a[[1]] = 'python'
a[250] = 'python'
```

```python
"""
10. 딕셔너리 값 추출하기
딕셔너리 a에서 'B'에 해당되는 값을 추출해보자.
"""
a = {'A':90, 'B':80, 'C':70}
result = a.pop('B')
print(a)  # {'A':90, 'C':70} 출력
print(result)  # 80 출력
```

```python
"""
11. 리스트에서 중복 제거하기
a 리스트에서 중복 숫자를 제거해보자.
"""
a = [1, 1, 2, 2, 3, 3, 3, 4, 4, 5]
aSet = set(a) # a리스트를 집합 자료형으로 변환 
b = list(aSet) # 집합 자료형을 리스트 자료형으로 다시 변환
print(b)  # [1, 2, 3, 4, 5] 출력
```

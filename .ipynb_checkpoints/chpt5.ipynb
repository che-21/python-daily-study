```python
class Fourcal:
		# pass 없애고 setdata 함수 정의
		def setdata(self, first, second):  # 메서드의 매개변수
				self.first = first             # 메서드의 수행문
				self.second = second           # 메서드의 수행문
				
a = Fourcal()
b = Fourcal()
a.setdata(4, 2)  # 이렇게 작성해도 setdata의 1번째 매개변수 self는 자동으로 전달됨

a.first
a.second

b.setdata(3, 7)
b.first

# a객체의 first 값은 b객체의 first 값에 영향받지 않고 원래 값을 유지
a.first
```

```python
class FourCal:
		# 연산할 숫자 입력
		def setdata(self, first, second):
				self.first = first
				self.second = second
		# 더하기
		def add(self):
				result = self.first + self.second
				return result
		# 곱하기
		def mul(self):
				result = self.first * self.second
				return result
		# 빼기
		def sub(self):
				result = self.first - self.second
				return result
		# 나누기
		def div(self):
				result = self.first / self.second
				return result
				
a = FourCal()
a.setdata(4,2)
a.add()
a.mul()
a.sub()
a.div()
```

```python
class Human:
	""" Super Class """
	# class 속성
	name = '이름'
	age = '나이'
	city = '도시'
	
	def show(self):
			print('Human 클래스의 메소드입니다.')

class Student(Human):
	""" Child Class """
	def __init__(self,name):
			self.name = name
	
	def show_name(self):
			print('이름은 ', self.name)
			
	def show_age(self):
			print('나이는 ', self.age)
			
a = Student('Jane')
a.show()  # 메소드 상속
a.show_age() # 속성 상속
a.show_name() # 자식노드에서 속성을 변경
```

```python
# 다중상속
class Animal:
	def eat(self):
			print("먹는중")
			
class Flyable:
	def fly(self):
			print("하늘을 나는 중")
			
class Bird(Animal, Flyable):
		pass
		
b = Bird()
b.eat()  # 먹는중
b.fly()  # 하늘을 나는 중
```

```python
# ValueError 뒤에 오류 메시지 적어줘야 함
# __str__ 메서드는 문자열을 변환함
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if self.amount <= 0:
            return ValueError

    def withdraw(self, amount):
        if self.amount < amount:
            return ValueError

total_accounts = BankAccount()

# 테스트 코드
a1 = BankAccount("홍길동", 5000)
a2 = BankAccount("김철수")
a1.deposit(5000)
print(a1)  # 홍길동님의 계좌: 잔액 10000원
print(BankAccount.total_accounts)  # 2
```

```python
class BankAccount:
    total_accounts = 0

    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        BankAccount.total_accounts += 1

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("입금액은 0보다 커야 합니다.")
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError("잔액이 부족합니다.")
        self.balance -= amount

    def __str__(self):
        return f"{self.owner}님의 계좌: 잔액 {self.balance}원"


# 테스트
a1 = BankAccount("홍길동", 5000)
a2 = BankAccount("김철수")
a1.deposit(5000)
print(a1)                          # 홍길동님의 계좌: 잔액 10000원
print(BankAccount.total_accounts)  # 2
```

```python
# 라이브러리 불러오기
import pandas as pd

# k:v 구조를 갖는 딕셔너리를 만들고 변수 dict_data 저장
dict_data = {'a': 1, 'b': 2, 'c': 3}

#판다스 Series() 함수로 딕셔너리(dict_data)를 시리즈로 변환, 변수 sr에 저장
sr = pd.Series(dict_data)

# 변수 sr의 자료형 출력
print(type(sr))

# 변수 sr에 저장되어 있는 시리즈 객체를 출력
print(sr)
```

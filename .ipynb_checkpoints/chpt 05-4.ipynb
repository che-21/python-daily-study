```python
# 05-4 예외처리
## 오류 일부러 발생 시키기
class Bird:
    def fly(self):
        raise NotImplementedError

class Eagle(Bird):
    pass

"""
# fly()를 오버라이딩하여 구현하지 않아 에러 발생
eagle = Eagle()
eagle.fly()
"""
class Eagle(Bird):
    def fly(self):
        print("very fast")

eagle = Eagle()
eagle.fly()
```

```python
## 예외 만들기 - 에러 발생
class MyError(Exception):
    pass

def say_nick(nick):
    if nick == '바보':
        raise MyError()
    print(nick)

say_nick("천사")
say_nick("바보")
```

```python
## 예외 만들기 - MyError 예외 처리
class MyError(Exception):
    pass

def say_nick(nick):
    if nick == '바보':
        raise MyError()
    print(nick)

try:
    say_nick("천사")
    say_nick("바보")
except MyError:
    print("허용되지 않는 별명입니다.")
```
